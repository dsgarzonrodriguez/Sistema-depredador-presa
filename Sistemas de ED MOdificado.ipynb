```python
import numpy as np
import matplotlib.pyplot as plt
```

* Parámetros

```python
# Parámetros del modelo

alfa  = 2
beta  = 1.2
gamma = 1
delta = 0.9

# Tamaño del paso
h = 0.01

# iteraciones
t_final = 16
```

* Condiciones iniciales

```python
R_0 = 1.0
F_0 = 0.5
```

```python
R_2_0 = 1.1
F_2_0 = 1.6666666
```

* Parámetros del método

```python
t = np.arange(0, t_final + h, h)
```

* Vectores para almacenar la solución

```python
R = np.zeros(len(t))
F = np.zeros(len(t))

R[0] = R_0
F[0] = F_0
```

```python
R_2 = np.zeros(len(t))
F_2= np.zeros(len(t))

R_2[0] = R_2_0
F_2[0] = F_2_0
```

* Método de Euler

```python
for n in range(len(t) - 1):

    # Derivadas en el punto actual
    dR = alfa*R[n] - beta*R[n]*F[n]
    dF = -gamma*F[n] + delta*R[n]*F[n]

    # Siguiente punto
    R[n+1] = R[n] + h*dR
    F[n+1] = F[n] + h*dF
```

```python
for n in range(len(t) - 1):

    # Derivadas en el punto actual
    dR_2 = alfa*R_2[n] - beta*R_2[n]*F_2[n]
    dF_2 = -gamma*F_2[n] + delta*R_2[n]*F_2[n]

    # Siguiente punto
    R_2[n+1] = R_2[n] + h*dR_2
    F_2[n+1] = F_2[n] + h*dF_2
```

* Gráficas R(t) y F(t)

```python
plt.figure(figsize=(10, 5))

plt.plot(t, R, label="Conejos R(t)")
plt.plot(t, F, label="Zorros F(t)")

plt.xlabel("t")
plt.ylabel("Población")
plt.title("Modelo depredador-presa — Método de Euler")
plt.legend()
plt.grid()

plt.show()
```

* Retrato de fase

```python
plt.figure(figsize=(10, 5))

plt.plot(R, F, label="Retrato de fase")

plt.xlabel("Conejos")
plt.ylabel("Lobos")
plt.title("Modelo depredador-presa — Método de Euler")
plt.legend()
plt.grid()

plt.show()
```

```python
n = len(F)//3 - 25
```

**Para las flechas:**

- Calcular la dirección (derivadas/cambios entre puntos)

- Seleccionar solo algunos puntos para que no se sature de flechas (paso)


```python
lista = [3, 5, 7, 10, 16]; np.diff(lista); lista[1::2]
```

```python
# Diferencia entre valores seguidos del array
dx = np.diff(R[:n])
dy = np.diff(F[:n])

# Se escogen cada 10 datos
paso = 10
xs = R[:n][::paso]
ys = F[:n][::paso]
dxs = dx[::paso]
dys = dy[::paso]

# Se establece un valor acorde la variación para los colores
C = np.sqrt(xs**2 + ys**2)
```

```python
plt.figure(figsize=(10, 5))

#plt.plot(R[:n], F[:n], label="Retrato de fase")
plt.quiver(xs, ys, dxs, dys, C, angles="xy", scale=0.9, width=0.005, zorder=0.1)
plt.plot(R_2[:n], F_2[:n], label="Retrato de fase 2")

plt.xlabel("Conejos")
plt.ylabel("Lobos")
plt.title("Modelo depredador-presa — Método de Euler")
plt.legend()
plt.grid()

plt.show()
```
